## リスト1.1　行列の和・積・転置

````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[0, 1], [1, 0]])

print(A @ B)    # 行列積
print(B @ A)    # 順序を変えると結果も変わる
print(A * B)    # 要素ごとの積（行列積ではない）
print(A.T)      # 転置
````

**本書に載っている出力**
```
[[2 1]
 [4 3]]
[[3 4]
 [1 2]]
[[0 2]
 [3 0]]
[[1 3]
 [2 4]]
```

## リスト1.3　内積・ノルム・コサイン類似度

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))                 # 内積

x = np.array([3, -4])
print(np.linalg.norm(x, ord=1))     # L1ノルム
print(np.linalg.norm(x))            # L2ノルム（既定）

u = np.array([1, 0])
v = np.array([1, 1])
cos = u @ v / (np.linalg.norm(u) * np.linalg.norm(v))
print(round(cos, 4))                # コサイン類似度
````

**本書に載っている出力**
```
32
7.0
5.0
0.7071
```

## リスト1.4　行列式・逆行列と、掛けて単位行列になる確認

````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
print(round(np.linalg.det(A), 4))   # 行列式
Ainv = np.linalg.inv(A)
print(Ainv)                         # 逆行列
print(np.round(A @ Ainv, 4))        # 単位行列に戻るか確認
````

**本書に載っている出力**
```
-2.0
[[-2.   1. ]
 [ 1.5 -0.5]]
[[1. 0.]
 [0. 1.]]
```

## リスト1.5　対称行列の固有値・固有ベクトル（eigh）

````python
import numpy as np

C = np.array([[2, 1], [1, 2]])
vals, vecs = np.linalg.eigh(C)      # 対称行列用（固有値は昇順）
print(vals)
print(np.round(vecs, 4))            # 各「列」が固有ベクトル
````

**本書に載っている出力**
```
[1. 3.]
[[-0.7071  0.7071]
 [ 0.7071  0.7071]]
```

## リスト1.6　数値微分で f'(2) を確かめる

````python
def f(x):
    return x**3 - 3 * x

h = 1e-5
x = 2
slope = (f(x + h) - f(x - h)) / (2 * h)   # 中心差分
print(round(slope, 4))                    # f'(2) = 3*2**2 - 3
````

**本書に載っている出力**
```
9.0
```

## リスト1.7　偏微分を数値で確かめる（点 (2, 1)）

````python
def f(x, y):
    return x**2 * y + 3 * y

x, y, h = 2.0, 1.0, 1e-5
fx = (f(x + h, y) - f(x - h, y)) / (2 * h)
fy = (f(x, y + h) - f(x, y - h)) / (2 * h)
print(round(fx, 4), round(fy, 4))   # 2xy と x**2 + 3
````

**本書に載っている出力**
```
4.0 7.0
```

## リスト1.8　シグモイド関数とその微分

````python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

s = sigmoid(0.0)
print(s)
print(s * (1 - s))      # シグモイドの微分 σ(1-σ)
````

**本書に載っている出力**

```
0.25
```

## リスト1.9　勾配降下法を3回繰り返す

````python
# f(w) = (w - 3)**2 を勾配降下法で最小化する
w = 0.0
lr = 0.1                      # 学習率 η
for step in range(1, 4):
    grad = 2 * (w - 3)        # df/dw
    w = w - lr * grad
    print(step, round(w, 4))
````

**本書に載っている出力**
```
1 0.6
2 1.08
3 1.464
```

## リスト1.10　学習率を変えて5回更新したときの w

````python
# 学習率ごとに「最適値 w=3 からの距離」がどう変わるか
for lr in [0.1, 0.5, 1.0, 1.1]:
    w = 0.0
    for _ in range(5):
        w = w - lr * 2 * (w - 3)
    print(lr, round(w, 4))
````

## リスト1.11　二項分布の確率と、正規分布の ±zσ の確率

````python
import math

n, p, k = 5, 0.5, 2
prob = math.comb(n, k) * p**k * (1 - p)**(n - k)
print(prob)

for z in [1, 2, 3]:                 # 正規分布の ±zσ に入る確率
    print(z, round(math.erf(z / math.sqrt(2)), 4))
````

**本書に載っている出力**
```
0.3125
1 0.6827
2 0.9545
3 0.9973
```

## リスト1.12　ベイズの定理で事後確率を計算

````python
p_d = 0.01          # 有病率 P(病気)
p_pos_d = 0.99      # 感度 P(陽性|病気)
p_pos_not_d = 0.05  # 偽陽性率 P(陽性|病気でない)

p_pos = p_pos_d * p_d + p_pos_not_d * (1 - p_d)
posterior = p_pos_d * p_d / p_pos
print(round(p_pos, 4))
print(round(posterior, 4))
````

**本書に載っている出力**
```
0.0594
0.1667
```

## リスト1.13　平均・分散・標準偏差・不偏分散

## リスト1.16　集合の演算（和集合・共通部分・補集合）とド・モルガンの法則

````python
U = set(range(1, 11))                              # 全体集合 U
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}
print(sorted(A | B), sorted(A & B))                # 和集合・共通部分
print(sorted(U - A))                               # 絶対補 A^c（U から A を除く）
print(sorted(B - A))                               # 相対補 B − A（B の中で A に入らないもの）
print(sorted(U - (A | B)) == sorted((U - A) & (U - B)))    # ド・モルガンの法則
print(len(A | B) == len(A) + len(B) - len(A & B))          # 包除原理（要素数）
# U の要素が同じ確率で選ばれるとき、確率は「要素数の比」
print(len(A | B) / len(U), len(A & B) / len(B))            # P(A∪B)、P(A|B) = P(A∩B)/P(B) = |A∩B|/|B|
````

**本書に載っている出力**
```
[1, 2, 3, 4, 5, 6, 7] [4, 5]
[6, 7, 8, 9, 10]
[6, 7]
True
True
0.7 0.5
```
