## リスト2.4　引数・戻り値・lambda

````python
def stats(xs, scale=1.0):
    total = sum(xs)
    mean = total / len(xs)
    return total * scale, mean      # 複数の値はタプルで返る

print(stats([1, 2, 3]))
t, m = stats([1, 2, 3], scale=2)
print(t, m)

def add_all(*args, **kwargs):
    return sum(args), kwargs

print(add_all(1, 2, 3, lr=0.1))
square = lambda x: x ** 2
print(list(map(square, [1, 2, 3])))
````

**本書に載っている出力**
```
(6.0, 2.0)
12 2.0
(6, {'lr': 0.1})
[1, 4, 9]
```

## リスト2.5　ミュータブルなデフォルト引数

````python
def add_item(x, items=[]):
    items.append(x)
    return items

print(add_item(1))
print(add_item(2))       # 前回の続きになる！

def add_item_safe(x, items=None):
    if items is None:
        items = []
    items.append(x)
    return items

print(add_item_safe(1))
print(add_item_safe(2))
````

**本書に載っている出力**
```
[1]
[1, 2]
[1]
[2]
```

## リスト2.6　fit / transform 型のクラス

````python
class MinMaxScaler:
    # 最小値0・最大値1にそろえる（scikit-learn と同じ fit / transform の形）
    def __init__(self):
        self.lo = None
        self.hi = None

    def fit(self, xs):
        self.lo = min(xs)
        self.hi = max(xs)
        return self

    def transform(self, xs):
        return [(x - self.lo) / (self.hi - self.lo) for x in xs]

sc = MinMaxScaler().fit([10, 20, 40])
print(sc.lo, sc.hi)
print(sc.transform([10, 20, 40]))
print(sc.transform([30]))        # 学習データの lo, hi を使う
````

**本書に載っている出力**
```
10 40
[0.0, 0.3333333333333333, 1.0]
[0.6666666666666666]
```

## リスト2.7　try / except / else / finally

````python
def safe_div(a, b):
    try:
        r = a / b
    except ZeroDivisionError:
        print("0で割れません")
        return None
    except TypeError as e:
        print("型エラー:", e)
        return None
    else:
        print("成功")
        return r
    finally:
        print("終了")

print(safe_div(6, 3))
print(safe_div(1, 0))
print(safe_div("a", 2))
````

**本書に載っている出力**
```
成功
終了
2.0
0で割れません
終了
None
型エラー: unsupported operand type(s) for /: 'str' and 'int'
終了
None
```

## リスト2.17　solve と lstsq

````python
import numpy as np

A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 10.])
x = np.linalg.solve(A, b)           # 連立方程式 Ax = b を解く
print(x)
print(np.allclose(A @ x, b))
print(round(np.linalg.det(A), 4), np.linalg.matrix_rank(A))

# 最小二乗法：点 (1,2), (2,3), (3,5) に直線 y = w0 + w1 x を当てはめる
X = np.array([[1., 1.], [1., 2.], [1., 3.]])   # 1の列（切片用）と x
y = np.array([2., 3., 5.])
w, res, rank, sv = np.linalg.lstsq(X, y, rcond=None)
print(np.round(w, 4))
````

## リスト2.23　groupby・value_counts・pivot_table

````python
import pandas as pd

sales = pd.DataFrame({
    "shop": ["A", "A", "B", "B", "B"],
    "item": ["x", "y", "x", "x", "y"],
    "qty": [3, 1, 2, 4, 5],
    "price": [100, 200, 100, 100, 200],
})
sales["amount"] = sales["qty"] * sales["price"]
print(sales.groupby("shop")["amount"].sum())
print(sales.groupby("shop").agg(qty_sum=("qty", "sum"), amount_mean=("amount", "mean")))
print(sales["item"].value_counts())
print(sales.pivot_table(index="shop", columns="item", values="qty",
                        aggfunc="sum", fill_value=0))
````

## リスト2.26　Seaborn で 4 種類のグラフを 1 枚に描く（相関行列つき）

````python
import matplotlib
matplotlib.use("Agg")                              # 画面を使わず、ファイルに描く設定
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

rng = np.random.default_rng(0)
x = rng.normal(0, 1, 200)
df = pd.DataFrame({"x": x, "y": 2 * x + rng.normal(0, 1, 200), "g": rng.choice(["A", "B"], 200)})
print(df.corr(numeric_only=True).round(2))         # 相関行列（heatmap の元データ）

fig, ax = plt.subplots(2, 2, figsize=(8, 6))
sns.histplot(df, x="x", bins=20, ax=ax[0, 0])                      # ヒストグラム
sns.scatterplot(df, x="x", y="y", hue="g", ax=ax[0, 1])            # 散布図（色分け）
sns.boxplot(df, x="g", y="y", ax=ax[1, 0])                         # 箱ひげ図
sns.heatmap(df[["x", "y"]].corr(), annot=True, ax=ax[1, 1])        # 相関のヒートマップ
fig.savefig("plots.png", dpi=100)
import os
print(os.path.getsize("plots.png") > 0)
````

**本書に載っている出力**
```
      x     y
x  1.00  0.88
```
